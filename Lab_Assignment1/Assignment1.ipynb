```python
# Merge the studentHomerooms with the merged dataset & name it df_studentInfo. Make sure that the key you are joining on are the same typecast for both datasets.
# Also ensure that there are no redundant columns, and finally display the first few rows.
# Hint: try joining on more than one column to avoid having to drop redundant columns, ie. on=['studentID', '<ColumnA>', '<ColumnB>'] ...
# (5 marks)

#Checking the column of both datasets
print(df_studentGrades.columns)
print(df_homeroom.columns)

# Making sure that studentID is the same type in both datasets
df_studentGrades['studentID'] = df_studentGrades['studentID'].astype(str)
df_homeroom['studentID'] = df_homeroom['studentID'].astype(str)

# Merging the datasets on studentID
df_studentInfo = pd.merge(
    df_studentGrades,
    df_homeroom,
    left_on=['studentID', 'firstName_English', 'lastName_English'],
    right_on=['studentID', 'firstName', 'lastName'],
    how='inner'
)
# To avoid dropping unnecessary columns, trying to join on multiple columns.
# Only keep essential columns (optional renaming)
df_studentInfo = df_studentInfo[['firstName_English', 'lastName_English', 'studentID',
                                 'grade_English', 'grade_Math', 'GPA', 'homeRoom']]

# Display first few rows
display(df_studentInfo.head())

```

```python
# The code below calculates the mean of each homeroom class.
df_grouped = df_studentInfo.groupby('homeRoom')['GPA'].mean()

# Printing the mean GPA per homeroom
print(df_grouped)

# Which homeroom has a higher GPA?
# (1 mark)
# Identify the homeroom with the highest mean GPA
highest_gpa_homeroom = df_grouped.idxmax()
print(f"Homeroom with the highest GPA: {highest_gpa_homeroom}")
```

```python
# Now let's work with multiple sorts. The .sort_values() method can sort by multiple columns, meaning you can specify a primary column and a secondary column for sorting.
# For example, by=['ColumnA', 'ColumnB'] will first sort by ColumnA and then by ColumnB.
# Create and print a DataFrame, df_studentsGPASorted_homeroom, which is sorted by homeRoom and then by GPA, both in descending order.
# (3 mark)

# Sorting the DataFrame by homeRoom and then by GPA, in descending order
df_studentsGPASorted_homeroom = df_studentInfo.sort_values(
    # primary: homeRoom, secondary: GPA
    by=['homeRoom', 'GPA'],
    # descending for both columns
    ascending=[False, False]
    # descending for both columns
 )

# printing the sorted DataFrame
print(df_studentsGPASorted_homeroom)
```

```python
# Do you notice a problem with the indices? They are basically meaningless now that we have sorted by homeroom and GPA.

# The code below changes the index of the dataframe so that it is more intuitive.
df_studentsGPASorted_homeroom['studentRank'] = df_studentsGPASorted_homeroom.groupby('homeRoom').cumcount() + 1
df_studentsGPASorted_homeroom['newIndex'] = df_studentsGPASorted_homeroom['homeRoom'].astype(str) + '-' + df_studentsGPASorted_homeroom['studentRank'].astype(str)
df_studentsGPASorted_homeroom = df_studentsGPASorted_homeroom.set_index('newIndex')
df_studentsGPASorted_homeroom = df_studentsGPASorted_homeroom.drop(columns=['studentRank'])

print(df_studentsGPASorted_homeroom)
```

```python
# step 5
# define a curve fitting function named fit_polynomial_curve
# the function should implement the linear least square estimation
# the function should include all important intermediate steps in LSE
# the input and output of the funciton are described in the comment below in red.
# please keep code clean and well-commented
# (7 marks)

def fit_polynomial_curve(a: np.ndarray, l: np.ndarray, P: np.ndarray, deg: int) -> tuple:
    '''
    Args:
        a:              function input (constants)
        l:              noisy observations
        P:              observation weight matrix
        deg:            degree of polynomial to fit to data
    Returns:
        x:              estimated polynomial coefficients
        cov_x:          uncertainty in estimated coefficients
        apv:            a posterior variance factor (goodness of fit)
        v:              observation residuals
    '''

  # ----- Step 1: Build design matrix A -----
    # A has columns [1, a, a^2, ..., a^deg]
    A = np.vander(a, N=deg+1, increasing=True) #function in NumPy creates a Vandermonde matrix

    # ----- Step 2: Using LSE method -----
    # x = (A^T P A)^(-1) A^T P l
    AtPA = A.T @ P @ A
    AtPA_inv = np.linalg.pinv(AtPA)
    x = AtPA_inv @ (A.T @ P @ l)

    # ----- Step 3: Compute residuals -----
    # v = l - A x
    v = l - A @ x

    # ----- Step 4: A Posterior Variance Factors  -----
    # apv = v^T P v / (n - u)
    n = len(l)          # number of observations
    u = deg + 1         # number of unknown parameters
    apv = (v.T @ P @ v) / (n - u)

    # ----- Step 5: Covariance of estimated parameters -----
    # Cov(x) = apv * (A^T P A)^(-1)
    cov_x = apv * AtPA_inv

    return (x, cov_x, apv, v)
```

```python
# step 6
# call the fit_polynomial_curve function and fetch the results
# use polynomial degrees of: 2, 4, 6
# (1 marks)
deg_list = [2, 4, 6]

# store results
results_uw = {}  # uw = unweighted

for deg in deg_list:
    x, cov_x, apv, v = fit_polynomial_curve(a, l, P, deg)
    results_uw[deg] = (x, cov_x, apv, v)
    print(f"Degree {deg} fit complete.")
```

```python
# step 10
# compute and print the norm between the predicted observations (output from each polynomial function) and the true observations
# (3 marks)

x = df_data['l'].values
y_true = df_data['l_true'].values

deg_list = [2, 4, 6]
norms = []

for deg in deg_list:
    # Create polynomial features manually
    X_poly = np.vstack([x**i for i in range(deg + 1)]).T

    # Solve for coefficients using least squares
    coeffs = np.linalg.lstsq(X_poly, y_true, rcond=None)[0]

    # Predict
    y_pred = X_poly @ coeffs

    # Compute L2 norm
    norm = np.linalg.norm(y_pred - y_true)
    norms.append(norm)
    print(f"Degree {deg}: L2 norm = {norm:.3f}")
```

```python
# step 12
# now, let's define the P matrix with the proper weights of observation
# use the variance of observation l_var, define P matrix
# (1 mark)
w = 1 / df_data['l_var'].values  # weights = 1 / variance
Proper = np.diag(w)              # diagonal weight matrix
```

```python
# step 13
# repeat the steps 6-10, using the proper P matrix
# (7 marks)

#Repeat Step 6
deg_list = [2, 4, 6]

# store results
results_w = {}  # w = weighted

for deg in deg_list:
    x, cov_x, apv, v = fit_polynomial_curve(a, l, Proper, deg)
    results_w[deg] = (x, cov_x, apv, v)
    print(f"Degree {deg} fit complete.")

#Repeat Step 7
# degree 2 polynomial (quadratic)
quadratic_w = np.polyval(results_w[2][0][::-1], a)

# degree 4 polynomial (quartic)
quartic_w = np.polyval(results_w[4][0][::-1], a)

# degree 6 polynomial (hexic)
hexic_w = np.polyval(results_w[6][0][::-1], a)

#Repeat Step 8
fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(14, 5))

# plot true observations
ax.plot(a, l_true, label="True f(a)", color='k')

# plot noisy observations
ax.scatter(a, l, label="Noisy f(a) + e", color='tab:blue', marker='o', alpha=0.3)

# plot polynomial fits
ax.plot(a, quadratic_w, label="Degree 2 (Quadratic)", color='orange')
ax.plot(a, quartic_w, label="Degree 4 (Quartic)", color='green')
ax.plot(a, hexic_w, label="Degree 6 (Hexic)", color='red')

# labels and title
ax.set_xlabel("a (radians)")
ax.set_ylabel("Function values")
ax.set_title("Polynomial Curve Fitting – Weighted LSE")

#reset range of x-axis to be -3 to 3, since if we use -1 to 1, it will not properly show the curve fitting as it will look like just a straight line
ax.set_xlim([-3, 3])

plt.show()

# legend
ax.legend()

# grid
ax.grid(True, which='both', ls='-', color='0.65')

fig.tight_layout()
plt.show()

#Repeat Step 9

print("A Posterior Variance Factors (Weighted LSE):")
for deg in results_w:
    print(f"Degree {deg}: {results_w[deg][2]:.3f}")

#Repeat Step 10
x = df_data['l'].values
y_true = df_data['l_true'].values
deg_list = [2, 4, 6]
weighted_norms = []

print("Weighted L2 Norms:")
for deg in deg_list:
    # Create polynomial features
    X_poly = np.vstack([x**i for i in range(deg + 1)]).T

    # Apply weighted least squares: β = (Xᵀ P X)⁻¹ Xᵀ P y
    XT_P = X_poly.T * w  # equivalent to Xᵀ @ P
    beta = np.linalg.inv(XT_P @ X_poly) @ (XT_P @ y_true)

    # Predict
    y_pred = X_poly @ beta

    # Compute weighted L2 norm: sqrt((y_pred - y_true)ᵀ P (y_pred - y_true))
    residual = y_pred - y_true
    norm = np.sqrt(residual.T @ P @ residual)
    weighted_norms.append(norm)
    print(f"Degree {deg}:= {norm:.3f}")
```
